```python
import torch
```

```python
#task1
tensor_A = torch.rand(7, 7)
print("Tensor A (7x7):\n", tensor_A)
tensor_B = torch.rand(1, 7)
print("\nTensor B (1x7):\n", tensor_B)

matmul_result = torch.matmul(tensor_A, tensor_B.T)
print("\nMatrix Multiplication Result (7x1):\n", matmul_result)
print("Result Shape:", matmul_result.shape)

torch.manual_seed(0)
tensor_C = torch.rand(7, 7)
tensor_D = torch.rand(1, 7)


matmul_seeded = torch.matmul(tensor_C, tensor_D.T)
print("\nMatrix Multiplication with Seed (7x1):\n", matmul_seeded)
print("Result Shape:", matmul_seeded.shape)

if torch.cuda.is_available():
    tensor_gpu = tensor_C.to("cuda")
    print("\nTensor moved to GPU ✅")
else:
    print("\nCUDA (GPU) not available on this device ")
```

```python
#task4
torch.manual_seed(0)

A = torch.rand(7, 7)
B = torch.rand(7, 7)

tensor_C = torch.mm(A, B)

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
print("\nResult (A x B):\n", tensor_C)
print("\nShape of Result:", tensor_C.shape)
```

```python
# task5
first_two_rows = tensor_C[:2]

print("First two rows of tensor_C:\n", first_two_rows)
print("Shape:", first_two_rows.shape)
```

```python
# task6
last_three_cols = tensor_C[:, -3:]

print("Last three columns of tensor_C:\n", last_three_cols)
print("Shape:", last_three_cols.shape)
```

```python
#task7
first_column = tensor_C[:, 0]

print("First column of tensor_C:\n", first_column)
print("Shape:", first_column.shape)
```

```python
#task8
reshaped_tensor = tensor_C.reshape(1, 49)

print("Reshaped tensor:\n", reshaped_tensor)
print("New shape:", reshaped_tensor.shape)
```

```python
#task9
concatenated = torch.cat((tensor_C, tensor_C), dim=1)

print("Concatenated tensor:\n", concatenated)
print("New shape:", concatenated.shape)
```

```python
# task10
max_value = torch.max(tensor_C)
min_value = torch.min(tensor_C)

print("Maximum value in tensor_C:", max_value.item())
print("Minimum value in tensor_C:", min_value.item())
```
